2가지 문제를 생각해 볼 수 있을 것 같습니다.

* **pdf 텍스트** : 각종 서식 태그와 한글 문서식 글머리 기호, 머리말과 담당자 정보 등이 담겨 있습니다.

8대공정 텍스트 문제부터 해결해 보겠습니다.

```python
# 모든 파일에 공통으로 들어간 블로그 메뉴/태그 텍스트
NOISE_LINES = {
    "8대 공정", "8대 공정 반도체+",
    "페이스북", "X", "카카오톡", "이메일",
    "관련태그", "‘8대 공정’, ‘8대공정’",
}

def clean_text(text):
    cleaned = []
    for line in text.splitlines():
        line = line.strip()
        if line == "" or line in NOISE_LINES:  # 빈 줄, 메뉴 텍스트 제거
            continue
        if line.startswith("☞"):  # 다른 글로 가는 링크 목록 제거
            continue
        if line in cleaned:  # 이미 나온 줄(중복 제목) 제거
            continue
        cleaned.append(line)
    return "\n".join(cleaned)
```

```python
# 저장 폴더와 저장 함수 준비
chunk_dir = output_dir / "chunking"
chunk_dir.mkdir(exist_ok=True)

output_dir = Path("output")
clean_md_path = output_dir / "clean_0911(12조간)반도체과, 반도체 첨단패키징 기술강국 도약 시동.md"

clean_md = clean_md_path.read_text(encoding="utf-8")

print(f"불러온 글자 수: {len(clean_md)}자")
print(clean_md[:300])

def save_chunks(chunks, name):
    lines = []
    for i, chunk in enumerate(chunks, 1):
        lines.append(f"===== 청크 {i} ({len(chunk)}자) =====")
        lines.append(chunk + "\n")

    (chunk_dir / f"{name}.txt").write_text("\n".join(lines), encoding="utf-8")
    print(f"{name}: 청크 {len(chunks)}개 → {chunk_dir / name}.txt")
```

### 텍스트 임베딩

텍스트 임베딩을 수행해 보겠습니다. 임베딩은 Sparse Vector와 Dense Vector 2가지가 존재합니다.

#### Sparse Vector

Sparse vector는 키워드 매칭 기반으로 관련 문서를 탐색하는 방법입니다. 가장 많이 사용되는 알고리즘은 BM25 알고리즘입니다.

우선 단어들을 한국어 형태소 분석기를 이용해 각 문서들에서 중요 키워드 단어들만 추출합니다. 예를 들어 "웨이퍼는"과 "웨이퍼를" 에서 똑같이 "웨이퍼"라는 키워드만 추출하는 식입니다.

```python
from kiwipiepy import Kiwi

total_chunks = pdf_chunks + text_chunks
print(f"PDF {len(pdf_chunks)}개 + txt {len(text_chunks)}개 = {len(total_chunks)}개")

kiwi = Kiwi()
kiwi.add_user_word("수율", "NNG")  # 분석기가 모르는 도메인 용어를 사전에 추가

def tokenize(text):
    # 명사(NN), 영어(SL), 숫자(SN)만 남기고 조사·어미는 버립니다.
    tokens = kiwi.tokenize(text)
    return [t.form for t in tokens if t.tag.startswith(("NN", "SL", "SN"))]

print(tokenize("웨이퍼는 반도체 집적회로의 핵심 재료입니다. EDS 공정에서 수율을 높인다."))
```

이제 토큰화된 문장을 BM25를 이용해 sparse vector로 변환합니다.

```python
from rank_bm25 import BM25Okapi

tokenized_chunks = [tokenize(chunk) for chunk in total_chunks]
bm25 = BM25Okapi(tokenized_chunks)

print(f"전체 단어 수: {len(bm25.idf)}개")

def to_sparse_vector(i):
    tf = bm25.doc_freqs[i]   # 이 청크에서 각 단어가 나온 횟수
    dl = bm25.doc_len[i]     # 이 청크의 길이
    vector = {}
    for term, freq in tf.items():
        weight = bm25.idf[term] * freq * (bm25.k1 + 1) / (
            freq + bm25.k1 * (1 - bm25.b + bm25.b * dl / bm25.avgdl))
        vector[term] = round(weight, 3)
    return vector

sparse_vectors = [to_sparse_vector(i) for i in range(len(total_chunks))]

# 7번째 txt 청크(웨이퍼 문서)에서 가중치가 높은 단어 상위 8개
# 가중치가 높은 단어 = 해당 문서에서만 많이 등장하는 단어를 의미
top = sorted(sparse_vectors[len(pdf_chunks)].items(), key=lambda x: -x[1])[:8]
print(top)
```

* 문서에 있는 키워드를 사용하지 않고 같은 의미를 다른 단어로 바꿔서 질문
* 내가 이 챗봇 서비스를 이용한다고 했을 때, 어떤 식으로 물어볼 지를 생각해서 자연스럽게 질문

[예시 문장]
* "첨단패키징 사업 예산은 얼마야?"
* "EDS 공정에서 불량 칩은 어떻게 걸러내?"

```python
# 👇 이 문장을 자유롭게 바꿔서 실행해 보세요!
query = "EDS 공정에서 불량 칩은 어떻게 걸러내나요?"

def search(query, k=3):
    query_tokens = tokenize(query)
    scores = bm25.get_scores(query_tokens)
    top_indices = sorted(range(len(scores)), key=lambda i: scores[i], reverse=True)[:k]

    print(f"질문: {query}")
    print(f"질문 토큰: {query_tokens}\n")
    for rank, i in enumerate(top_indices, 1):
        matched = [t for t in set(query_tokens) if t in tokenized_chunks[i]]
        preview = total_chunks[i][:80].replace("\n", " ")
        print(f"{rank}위 | 청크 {i} | 점수 {scores[i]:.2f} | 겹친 단어 {matched}")
        print(f"     {preview}\n")

search(query)
```

다음은 Dense Vector 입니다. Dense Vector는 딥러닝으로 학습된 임베딩 모델을 사용합니다. 

Sparse Vector와의 차이점은 Dense Vector는 보다 의미적 유사도를 볼 수 있다는 것으로 키워드가 겹치지 않더라도 비슷한 의미를 가진 문서를 찾아올 수 있다는 장점이 있습니다.

여기선 chatgpt의 api를 활용해 임베딩을 수행합니다.

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_openai import OpenAIEmbeddings

load_dotenv(find_dotenv())

embedding_model = OpenAIEmbeddings(
    model="text-embedding-3-small",
    base_url=os.getenv("Embedding_BASE_URL"),
)
```

이제 임베딩을 수행해 보겠습니다.

```python
# 16개 청크를 한 번에 임베딩 API로 보내 벡터로 바꿉니다.
dense_vectors = embedding_model.embed_documents(total_chunks)

print(f"벡터 개수: {len(dense_vectors)}개")
print(f"벡터 하나의 길이: {len(dense_vectors[0])}차원")
print(f"첫 번째 벡터 앞부분: {[round(v, 4) for v in dense_vectors[0][:8]]}")
```

Dense vector 기반의 검색도 실험해 보도록 하겠습니다. 아래 문장을 바꿔 가면서 직접 관련 문서를 잘 찾아오는지 직접 확인해 보세요.

키워드가 겹치지 않는 유사 문서를 잘 찾아오는지를 위주로 직접 확인해 보세요.

```python
import numpy as np

# 👇 이 문장을 자유롭게 바꿔서 실행해 보세요!
query = "첨단패키징 사업 예산은 얼마야?"

def dense_search(query, k=3):
    query_vector = np.array(embedding_model.embed_query(query))
    doc_vectors = np.array(dense_vectors)

    # 코사인 유사도: 두 벡터의 방향이 비슷할수록 1에 가까워집니다.
    scores = doc_vectors @ query_vector / (
        np.linalg.norm(doc_vectors, axis=1) * np.linalg.norm(query_vector))
    top_indices = np.argsort(scores)[::-1][:k]

    print(f"질문: {query}\n")
    for rank, i in enumerate(top_indices, 1):
        preview = total_chunks[i][:80].replace("\n", " ")
        print(f"{rank}위 | 청크 {i} | 유사도 {scores[i]:.3f}")
        print(f"     {preview}\n")

dense_search(query)
```

보통은 둘의 장단점이 명확하고 상호보완적인 관계를 갖기 때문에 두 방법을 섞어서 사용합니다. 두 방식을 섞은 hybrid 방식으로 retriever를 구성해 보겠습니다.

```python
from langchain_community.retrievers import BM25Retriever
from langchain_community.vectorstores import FAISS
from langchain_classic.retrievers import EnsembleRetriever

metadatas = [{"chunk_id": i} for i in range(len(total_chunks))]

# 1) 키워드 검색기: 앞에서 만든 한국어 tokenize 함수를 그대로 사용
bm25_retriever = BM25Retriever.from_texts(
    total_chunks, metadatas=metadatas, preprocess_func=tokenize, k=3)

# 2) 의미 검색기: 앞에서 만든 dense_vectors를 재사용 (API 재호출 없음)
vectorstore = FAISS.from_embeddings(
    list(zip(total_chunks, dense_vectors)), embedding=embedding_model, metadatas=metadatas)
dense_retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

# 3) 두 검색기를 섞은 하이브리드 검색기 (weights = [BM25 비중, Dense 비중])
hybrid_retriever = EnsembleRetriever(
    retrievers=[bm25_retriever, dense_retriever], weights=[0.5, 0.5])
```

마찬가지로 hybrid_retriever도 쿼리를 바꿔 가면서 직접 확인해 봅니다.

앞에서 잘 찾지 못했던 유사 문서들을 이번엔 잘 갖고 오는지 한번 확인해 봅니다.

```python
# 👇 질문과 비중을 자유롭게 바꿔서 실행해 보세요!
query = "첨단패키징 사업 예산은 얼마야?"
hybrid_retriever.weights = [0.5, 0.5]  # [BM25, Dense] 합이 1이 되도록

def hybrid_search(query):
    docs = hybrid_retriever.invoke(query)

    print(f"질문: {query}\n")
    for rank, doc in enumerate(docs, 1):
        preview = doc.page_content[:60].replace("\n", " ")
        print(f"{rank}위 | 청크 {doc.metadata['chunk_id']:>2} | {preview}")

hybrid_search(query)
```

이제 앞에서 만든 retriever를 활용해 rag 에이전트를 완성해 봅시다. 지금까지와 마찬가지로 tool로 만들고, create_agent()의 tool로 제공하면 됩니다.

```python
from langchain.tools import tool

@tool
def search_documents(query: str) -> str:
    """반도체 8대 공정 설명 자료와 반도체 첨단패키징 보도자료에서 질문과 관련된 문서를 검색합니다.

    query: 검색할 내용을 담은 짧은 문장이나 키워드 (예: "EDS 공정 불량 칩 판정")
    """
    docs = hybrid_retriever.invoke(query)

    results = []
    for doc in docs[:4]:  # 너무 많이 넘기지 않도록 상위 4개만 전달
        results.append(f"[청크 {doc.metadata['chunk_id']}]\n{doc.page_content}")
    return "\n\n---\n\n".join(results)
```

```python
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="gpt-4o-mini",
    base_url=os.getenv("gpt_4o_mini_BASE_URL"),
    temperature=0,
)

SYSTEM_PROMPT = """너는 반도체 공정과 산업 정책 자료를 안내하는 사내 어시스턴트다.
- 질문에 답하기 전에 반드시 search_documents 도구로 관련 문서를 검색해라.
- 검색된 문서에 있는 내용만으로 답하고, 문서에 없으면 "자료에서 찾을 수 없습니다"라고 답해라.
"""

rag_agent = create_agent(model, tools=[search_documents], system_prompt=SYSTEM_PROMPT)
```
